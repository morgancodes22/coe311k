## Numerical Methods Implementation

- Well-commented Python functions for:
    - Euler's Forward method
    - RK4 method
- Brief explanation of your implementation approach


```python
# Eulers Foward

def Euler(y_not, step, start, stop):

    # setup
    num_steps = int(round((stop - start) / step))
    x = np.linspace(start, stop, num_steps + 1)
    y = np.zeros(num_steps + 1)

    # initial condition
    x[0] = start
    y[0] = y_not

    # go through each step
    for i in range(num_steps):
        x[i+1] = x[i] + step 
        y[i+1] = y[i] + step * f(x[i], y[i])
    return x, y


def rk4(x0, y0, h, x_end):
    
    n = int((x_end - x0) / h) # number of steps
    x = x0
    y = y0
    
    for i in range(n): # fidning each section from the taylor series for every point
        k1 = h * f(x, y)
        k2 = h * f(x + 0.5*h, y + 0.5*k1)
        k3 = h * f(x + 0.5*h, y + 0.5*k2)
        k4 = h * f(x + h, y + k3)
        
        y += (1.0/6.0) * (k1 + 2*k2 + 2*k3 + k4)
        x += h
    return y
```
