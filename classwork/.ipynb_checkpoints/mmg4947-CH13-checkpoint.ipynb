```python
import matplotlib.pyplot as plt
import numpy as np

n = np.linspace(1, 100, 100)

plt.plot(n, np.log(n), label='O(log n)')
plt.plot(n, n, label='O(n)')
plt.plot(n, n * np.log(n), label='O(n log n)')
plt.plot(n, n**2, label='O(n^2)')
plt.plot(n, n**3, label='O(n^3)')
plt.plot(n, 2**n, label='O(2^n)')

plt.yscale('log')  # Logarithmic scale for better visualization
plt.title('Big O Complexity Growth Rates')
plt.xlabel('Input Size (n)')
plt.ylabel('Operations')
plt.legend()
plt.show()
```

# Challenge 13!

```python
import numpy as np
import time
import matplotlib.pyplot as plt

rng = np.random.default_rng()

def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C

def naive_time(A,B):
    start_time = time.perf_counter()
    naive_matrix_multiplication(A, B)
    end_time = time.perf_counter()
    return end_time - start_time


def numpy_time(A,B):
    start_time = time.perf_counter()
    np.dot(A, B)
    end_time = time.perf_counter()
    return end_time - start_time
```

```python

n=np.array([3,6,12,24,48,96,192,384])
naive_time_n = np.zeros(len(n))
numpy_time_n = np.zeros(len(n))
for i in range(len(n)):
    mat1 = rng.random(size=(n[i], n[i]))
    mat2 = rng.random(size=(n[i], n[i]))
    
    naive_time_n[i] = naive_time(mat1, mat2)
    numpy_time_n[i] = numpy_time(mat1, mat2)
    


matrix_num = n
plt.plot(matrix_num,naive_time_n, '-g', label="naive function")
plt.plot(matrix_num,numpy_time_n, '-b', label="numpy function")
plt.xlabel("matirx number")
plt.ylabel("runtime (seconds)")
plt.title("time complexity of matrix multiplcation")
plt.legend()
plt.show()
```

notes - 
the numpy function consistantly taking the same amoiunt of time to run no mater the matrix size but the naive function takes alot longer when the matrix is larger
