```python
import numpy as np
import matplotlib.pyplot as plt


# Define parameters
h_vals = np.array([0.1, 0.5, 1.0, 1.5]) # step size
#print(len(h_vals)) #debug
x_start, x_end = 0, 10

# Function representing dy/dx = -2y
def f(x, y):
    return -2 * y
def exact(x):
    return np.exp(-2*x)


def localError(x_n, step):
    y_exact_n = exact(x_n)
    y_exact_next = exact(x_n + step)
    y_aprox = y_exact_n + step * f(x_n, y_exact_n)
    return abs(y_exact_next - y_aprox)

def globalError(y_approx, x_n):
    return abs(exact(x_n) - y_approx)

# Implement Euler's method

for j in range(len(h_vals)): 
    #set up
    h = h_vals[j]
    num_steps = int((x_end - x_start) / h)
    
    x_values = np.zeros(num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    y_exact = np.zeros(num_steps + 1)
    local_error = np.zeros(num_steps + 1)
    global_error = np.zeros(num_steps + 1)
    
    # Initial condition
    x_values[0] = x_start
    y_values[0] = 1
    y_exact[0] = exact(x_start)
    global_error[0] = 0
    local_error[0] = 0
    
    for i in range(num_steps):
        
        step = min(h, x_end - x_values[i])
        
        x_values[i + 1] = x_values[i] + step
        y_values[i + 1] = y_values[i] + step * f(x_values[i], y_values[i])
        y_exact[i + 1] = exact(x_values[i + 1])
        
        local_error[i + 1] = localError(x_values[i], step)
        global_error[i + 1] = globalError(y_values[i + 1], x_values[i + 1])
    
    plt.subplot(2,2,1)    
    plt.plot(x_values, global_error, 'b--', label=f"global error h={h:.2f}")
    plt.plot(x_values, local_error, 'r--', label=f"local error h={h:.2f}")
    plt.legend()
    
    plt.subplot(2,2,2)
    plt.plot(x_values, y_values, 'b', label=f"aprox h={h:.2f}")
    plt.plot(x_values, y_exact, 'r', label=f"exact")
    
    plt.legend()
    plt.show()
```

# analysis
As h increases both the local and global error grow, as h decreases, the more the exact and aproximation track closley. When h=1 or 1.5 the gragh behaves wildly likley because it is out of Eulers Stability range
