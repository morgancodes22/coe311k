# matplotlib freeplay

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(-100,100,50)
```

```python
print(x)
```

```python
plt.subplot(2, 2, 1) # (row, column, active area)
plt.plot(x, x**2, 'r--')
plt.subplot(2,2,2)
plt.plot(x, np.sin(x), 'g')
plt.subplot(2,2,3)
plt.plot(x,np.sin(x), 'm')
plt.subplot(2,2,4)
plt.plot(x,x**2, 'k-.')
plt.show()
```
