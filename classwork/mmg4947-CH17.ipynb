```python
import numpy as np
import matplotlib.pyplot as plt


def euler_second(k, h, x_start=0, x_end=10, y_0=1.0, v_0=0.0):
    # Create arrays to store x, y, and v values
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))

    # Set initial conditions
    y_values[0] = y_0
    v_values[0] = v_0

    # Euler's Method loop
    for i in range(1, len(x_values)):
        # Current values
        x_n = x_values[i - 1]
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]

        # Update equations
        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)

    return x_values, y_values, v_values

#testing how h vals effects
k = 1
h_values = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]

plt.figure(figsize=(12, 6))

for h in h_values:
    x_values, y_values, v_values = euler_second(k, h)
    plt.plot(x_values, y_values, label=f"h = {h}")
    
plt.xlabel("x")
plt.ylabel("y")
plt.title("Euler's Method for second derivitive = -ky with different h values and k=1 ")
plt.legend()
plt.show()


# testing how diff k vals effect
h = 0.01
k_values = [0.5, 1, 2, 5, 10]

plt.figure(figsize=(12, 6))

for k in k_values:
    x_values, y_values, v_values = euler_second(k, h)
    plt.plot(x_values, y_values, label=f"k = {k}")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Euler's Method for second derivitive = -ky with different k values and h=0.01)")
plt.legend()
plt.show()


#specific case

k = 10
h = 0.01

x_values, y_values, v_values = euler_second(k, h)

plt.figure(figsize=(10, 5))
plt.plot(x_values, y_values, label="Position y(x)")
plt.plot(x_values, v_values, label="Velocity v(x)")
plt.xlabel("x")
plt.ylabel("y and v")
plt.title("Solution of second derivitive = -ky with k = 10 and h = 0.01")
plt.legend()
plt.grid(True)
plt.show()
```

For small step sizes like h=0.01 and h=0.02 the solution stays close to the normal oscillation. As h gets larger, the solution starts to lose its acuatecy more and more from what it should look like. at h=0.3 and h=0.5 the graph becomes unstable. the beginning of the graph still looks somewhat similar, the error keeps building as more steps are calculated, so the end of the graph looks much worse.

Changing k changes how fast the gragh oscillates. As k increases the number of oscillations or humps increases. these rapid changes in the derivitive makes eulers method less accurate unless h is very small.

so all in all the larger values of h or k make the method less accurate and worse. Eulers method does not work very well for this stiff equation when the step size is too large.
