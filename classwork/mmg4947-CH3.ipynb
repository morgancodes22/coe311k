# pi funtion

```python
import numpy as np
import matplotlib.pyplot as plt 
import random
import math 

def estPi(numPoints):
    
    count = 0 #counts what lands in the circle
    for i in range(0,numPoints):
        
        #get a random point
        x = random.uniform(-1,1)
        y = random.uniform(-1,1)
        
        #is it in the circle?
        if(x**2 + y**2 <1):
            count= count+1
    
    return 4.0*count/numPoints
```

```python
#run the funtion 100 times with the input of 1000 and save the results in results
results = np.empty(100) 
for i in range(0,100): 
    results[i] = estPi(1000)

#calculate errors into a array
trueError = abs(results - np.pi) 
reletiveError = abs( (results-np.pi)/ np.pi)
```

```python
#plot everything
x= np.arange(1,101,1)
plt.title("error")
plt.plot(x, trueError, 'r', label='true error')
plt.plot(x, reletiveError, 'g', label='reletive error')
plt.legend()
plt.show()
```
