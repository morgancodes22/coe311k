# Leibniz Series for Pi

```python
import numpy as np
import matplotlib.pyplot as plt 
import random
import math 
```

```python
nums = np.zeros(5000) #each term in series
error = np.zeros(50) #every 100th error
s = 0 #runing sum

for i in range(0,5000): #go through all the values of i to find the series
    nums[i]= ((-1)**i) / (2*i +1) # find the specific term
    s=s+nums[i]
    if((i+1) % 100 == 0): #calculate every 100th error
        #print(((i+1)//100)-1) #debug
        error[((i+1)//100)-1] = abs( (s*4 - np.pi)/ np.pi) #store the error to be plotted
        print(error[((i+1)//100) -1]) #print out every 100th error

#print(nums[0:5]) #debug
#print(np.sum(nums[0:10]) * 4) #debug
#print(error) #debug
```

```python
#plot results
x = np.arange(100, 5001, 100) 
plt.title("error")
plt.plot(x, error, 'g')
plt.show()
```

# Analyzing the gragh:
The more terms we added to the series, the less and less the error became because there was more digits in a infinite number like pi. this represents the Truncation Error. now the trick is deciding how to balance the errors and memory for the best performance on the particular task!
