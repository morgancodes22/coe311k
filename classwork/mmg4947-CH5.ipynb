## Singular Value Decomposition (SVD)
Singular Value Decomposition (SVD) is a factorization of a matrix into three matrices:

A=UΣV T

Where:
- U and V are orthogonal matrices - where the transpose is the same!
- Σ is a diagonal matrix of singular values

SVD is useful in many applications such as data compression, image processing, and dimensionality reduction

```python

# Define a matrix
A = np.array([[1, 2], [3, 4], [5, 6]])

# Perform SVD
U, sigma, Vt = np.linalg.svd(A)

print("U matrix:\n", U)
print("Sigma values:", sigma)
print("V^T matrix:\n", Vt)
```

# Gaussian Elimination, Python

Steps:
1. Represent the system in matrics form
2. Forward elimination 
    Identify the pivot element (The pivot element is the first non-zero element in the current column that hasn't been eliminated)
   -  Swap rows if the pivot element is 0
   - Normalize the pivot
   -  Eliminate lower elements
   -  Move to the next col;um
3. Backward Substitution
   -  Start with the Last Equation
   -  Substitute and Solve
   - Repeat this process, moving upwards row by row, substituting known values into previous equations.
   -  Continue Until All Variables Are Solved
4. Interpret the Results
   -  Check for Consistency


Rules:
1. You can multiply any row by a constant (other than zero).
2. You can swap any two rows.
3. You can add two rows together.


```python
import numpy as np

def gaussian_elimination(A, b):
    n = len(b)
    # Forward elimination
    for i in range(n):
        # Pivoting
        max_row = max(range(i, n), key=lambda x: abs(A[x][i]))
        A[i], A[max_row] = A[max_row], A[i]
        b[i], b[max_row] = b[max_row], b[i]
        
        # Make the rows below this one 0 in current column
        for j in range(i + 1, n):
            ratio = A[j][i] / A[i][i]
            A[j] = [A[j][k] - ratio * A[i][k] for k in range(n)]
            b[j] -= ratio * b[i]
    
    # Backward substitution
    x = [0] * n
    for i in range(n - 1, -1, -1):
        x[i] = (b[i] - sum(A[i][j] * x[j] for j in range(i + 1, n))) / A[i][i]
    return x
```

```python
# Example Usage
A = [[2, 5, -1], [4, -2, 2], [-2, 1, 2]]
b = [5, 6, -1]
solution = gaussian_elimination(A, b)
print(f"Solution: {solution}")
```

```python
#check solutions
ans = np.linalg.solve(A,b) #this does the exact same thing
print(ans)
```

# LU Decomposition

a mathematical method used to factor a given square matrix A into the product of two matrices: a lower triangular matrix L and an upper triangular matrix U. 
- helpful for solving systemss, inverting matrices, and compyting determinants

big idea: breaking a porblem up into 2 easier problems


A=LU

L - a lower triangulare matrix (everything above diag is 0)
U - a upper triangler (everything below diag is 0)


Ly=b
Ux=y


key concepts!!!
    -matrix factorization
    - Simplifies the computation of matrix inverses and determinants.
    -  LU decomposition with partial pivoting

## Steps

1. intialize L as a identiy matrix(all zeros and 1s) and U as a copy of A
2. for each column
    preform a elimination below the pivot element to gets 0s

The formulia is UGLYYYY but the algorithm is simple

```python
import numpy as np

def lu_decomposition(A):
    n = len(A)
    L = np.eye(n)  # Initialize L as an identity matrix
    U = np.array(A, dtype=float)  # Copy of A to avoid
    # modifying the original matrix
    # modifying the original matrix
    
    for i in range(n):
        # Eliminate elements below the pivot (U[i, i])
        for j in range(i+1, n):
            if U[i, i] == 0:
                raise ValueError("Zero pivot encountered. LU decomposition cannot proceed.")
        factor = U[j, i] / U[i, i]  # Multiplier for row
         # operation
        L[j, i] = factor  # Store multiplier in L

                # Update the rows of U
        for k in range(i, n):
                U[j, k] -= factor * U[i, k]

    return L, U
```

```python
# Example usage
A = np.array([[2, 3, 1],[4, 7, -1],[-2, 1, 2]])

L, U = lu_decomposition(A)

print("Matrix A:\n", A)
print("Lower Triangular Matrix L:\n", L)
print("Upper Triangular Matrix U:\n", U)
```

#### NOTE: LU decomposition isn't directly available in NumPy but can be computed using the scipy.linalg package).

```python
from scipy.linalg import lu

# Perform LU Decomposition
A = np.array([[2, 3, 1],
[4, 7, -1],
[-2, 1, 2]])

P, L, U = lu(A)
print("Matrix A:\n", A)
print("Permutation Matrix P:\n", P)
print("Lower Triangular Matrix L:\n", L)
print("Upper Triangular Matrix U:\n", U)
```
