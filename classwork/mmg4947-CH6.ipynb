# Iterative Methods

### Iterive methods!!!

idea: take a guess and compaire resluts and then refining the guess then repeat until the error is reasonably small

1. start with an initial guess - often all zeros or any resonable estimate
2. iterative update - updated based on previous iteration
3. convergence check - check if the solutions are simulare enough
4. stop

CONDITIONS:
- we want it to be diagonally dominate - the numbers in the diagonal to be bigger then anything in that row or column
        - can maniupuate with rules earler to make it this way
- symmetricly postively definite
    - producies positive numbers and close to merriorable


##### advantage: memory efficent, scaliblity, flexibility
##### disavantages : solver concergences based on starting guess, convergence depends on the matrix

## Jacobi Method

1. initialize to zero
2. update until x values based on hte last iteration until converges

```python


def jacobi(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    for k in range(max_iterations):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations
```

```python
A = np.array([[10, -1, 2, 0, 0],
              [-1, 11, -1, 3, 0],
              [2, -1, 10, -1, 0],
              [0, 3, -1, 8, -2],
              [0, 0, 0, -2, 9]])

b = np.array([14, 30, 26, 25, 37])

x0 = np.zeros_like(b)
tol = 1e-6
max_iterations = 100

timeStart =time.time()
print(timeStart)
solution, iterations = jacobi(A, b, x0, tol, max_iterations)
timeEnd = time.time()
print(timeEnd)

print("total time: ", (timeEnd-timeStart))
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
```

## Gauss-seidel

```python

def gauss_seidel(A, b, x0, tol, max_iterations):
    n = len(b)
    x = x0.copy()
    
    for k in range(max_iterations):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        # Check for convergence
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k
        x = x_new
    return x, max_iterations
```

```python
A = np.array([[10, -1, 2, 0, 0],
              [-1, 11, -1, 3, 0],
              [2, -1, 10, -1, 0],
              [0, 3, -1, 8, -2],
              [0, 0, 0, -2, 9]])

b = np.array([14, 30, 26, 25, 37])

x0 = np.zeros_like(b)
tol = 1e-6
max_iterations = 100

timeStart =time.time()
print(timeStart)
solution, iterations = gauss_seidel(A, b, x0, tol, max_iterations)
timeEnd = time.time()
print(timeEnd)

print("total time: ", (timeEnd-timeStart))
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
```

Both method converge, the jacobi used 1 iteration but took .001 time while Guass-Seidel used 2 iterations but took .00068 time. Gauss-Seidel is significantly faster in this case but did need another iteration.
