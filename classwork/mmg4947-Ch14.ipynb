```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function and the domain
f = lambda x: np.sin(x)
a, b = 0, 2 * np.pi  # Interval
N = 50  # Number of sample points

# Create a uniform grid
x = np.linspace(a, b, N)

h = (b - a) / (N - 1)

exact = np.cos(x)
central = (f(x + h) - f(x - h)) / (2 * h)
forward = (f(x + h) - f(x)) / h
backward = (f(x) - f(x - h)) / h

plt.subplot(2, 2, 1)
plt.plot(x, exact, 'o-')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title('Exact derivative - cos(x)')

plt.subplot(2, 2, 2)
plt.plot(x, central, 'o-')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title('Central Approx')

plt.subplot(2, 2, 3)
plt.plot(x, forward, 'o-')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title('Forward Approx')


plt.subplot(2, 2, 4)
plt.plot(x, backward, 'o-')
plt.xlabel('x')
plt.ylabel("f'(x)")
plt.title('Backward Approx')


plt.suptitle('Comparing cos(x) with diff Approx Methods')
plt.tight_layout(pad = 2.0)#had to google how to space things out
plt.show()
```
