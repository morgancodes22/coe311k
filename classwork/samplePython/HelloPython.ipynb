# Monte Carlo to fins a value of pi 

```python
import random
import math
def approxPi(numSamples):
    counter = 0
    for s in range(0,numSamples):
        x =  random.uniform(-1.0, 1.0)
        y = random.uniform(-1.0, 1.0)
        
        if( (x**2) + (y**2) <1):
            counter = counter+1
    return 4.0*counter/numSamples
```

```python
est = approxPi(10)
print(est)
```

```python
for i in range(1,6):
    print( approxPi(10**i))
    
#it should get more and mroe accurate as the number of sample increase
```
