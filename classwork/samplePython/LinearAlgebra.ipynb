Solving systems - y=mx+b
Where:
A is the coefficient matrix,
x is the vector of unknowns,
b is the constant vector.

2x + 3y = 5
4x - y = 3

```python
# Coefficient matrix A and constant vector b
A = np.array([[2, 3], [4, -1]]) #unknonwns: first row are x and second row is y
b = np.array([5, 3]) # the left hande sides!
# Solve the system A * x = b
x = np.linalg.solve(A, b)
print("Solution (x, y):", x)
```

## Eigenvalues and Eigenvectors
in a squar matrix A, there is a 𝜆, scalar and a vector, v where
𝐴𝑣 = 𝜆𝑣

Then 𝜆 is an eigenvalue of 𝐴, and 𝑣 is the corresponding eigenvector.

Eigenvectors are directions that don’t change direction when a matrix acts on them.
    

Eigenvalues tell you how much those directions get stretched or squished.


1. Solving systems of differential equations
- If eigenvalues are negative → system decays
- Positive → system grows
- Complex → oscillations
2. Data Science & Machine Learning (PCA)
- Eigenvectors = directions of maximum variance in data
- Eigenvalues = how important each direction is
3. Stability & convergence in Numerical Methods
- If spectral radius < 1 → converges
- Eigenvalues explain why some methods blow up and others behave nicely.
4. Vibrations & modes (physics & engineering)
- Eigenvectors = natural vibration modes
- Eigenvalues = frequencies
5. Image processing & computer graphics
- Eigenvectors capture dominant patterns in images.
6. Google & networks (PageRank!)
- Eigenvector = “importance” of a webpage
- Eigenvalue = how stable that ranking is
7.  Matrix powers & fast computation
- Diagonalizing a matrix = using eigenvalues to simplify hard problems.


### Eigenvectors are the “special directions” a system naturally likes to move in.
### Eigenvalues tell you whether motion in those directions grows, shrinks, or oscillates.


How to solve them in the Python???

```python
A = np.array( [[2,1], [1,2]])

eigenvalues, eigenvectors = np.linalg.eig(A) #eig returns the values and vectors
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

The inverse of a matrix A multiplied by the A is the identiry matrix

```python
# Define a matrix
A = np.array([[4, 7], [2, 6]])

# Compute the inverse of A
A_inv = np.linalg.inv(A)
print("Inverse of A:\n", A_inv)

# Verify: A * A_inv should be the identity matrix
identity = np.dot(A, A_inv)
print("A * A_inv:\n", identity)
```

## Singular Value Decomposition (SVD)
Singular Value Decomposition (SVD) is a factorization of a matrix into three matrices:

A=UΣV T

Where:
- U and V are orthogonal matrices - where the transpose is the same!
- Σ is a diagonal matrix of singular values

SVD is useful in many applications such as data compression, image processing, and dimensionality reduction

```python

# Define a matrix
A = np.array([[1, 2], [3, 4], [5, 6]])

# Perform SVD
U, sigma, Vt = np.linalg.svd(A)

print("U matrix:\n", U)
print("Sigma values:", sigma)
print("V^T matrix:\n", Vt)
```
