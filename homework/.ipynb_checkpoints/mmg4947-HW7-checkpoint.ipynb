Using Ulers method to solve dy/dx =-2y*cos(x), y(0) = 1

```python
import numpy as np
import matplotlib.pyplot as plt


# Function representing dy/dx = -2y*cos(x)
def f(x, y):
    return -2 * y * np.cos(x)

# Exact - y=e^(-2sin(x))
def exact(x):
    return np.exp(-2 * np.sin(x))

def Euler(y_not, step, start, stop):

    #setup
    num_steps = int(round((stop - start) / step))
    x = np.linspace(start, stop, num_steps + 1)
    y = np.zeros(num_steps + 1)

    # initial condition 
    x[0]= start
    y[0]= y_not

    # go through each step
    for i in range(num_steps):
        x[i+1]= x[i] + step
        y[i+1]= y[i] + step*f(x[i], y[i])  
    return x, y

h_vals = [0.01, 0.05, 0.15, 0.20, 0.25, 0.30, 0.35, 0.40, 0.45, 0.50, 0.55, 0.60, 0.65, 0.70, 0.75, 0.80, 0.85, 0.90, 0.95, 1.00]
avg_err= []
local_err_15= []

# subplots to keep organized
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

# Implement Euler's method for each h
for j in range(len(h_vals)):
    
    h = h_vals[j]
    x, y = Euler(1, h, 0, 10*np.pi) # find aprox
    
    #print('x:', x)
    #print('y:', y) #debug
    
    y_exact = exact(x)
    #print('y exact:', y_exact)
    
    
    globalErr = np.abs(y - y_exact)
    avg_err.append(np.mean(globalError))

    # error at 15
    i15 = np.argmin(np.abs(x - 15.0)) # setp size doesnt let it be perfect so round
    local_err_15.append(np.abs(y[i15] - exact(x[i15])))

    # choose what subplot it goes on
    plot_num = j // 5  # 5 lines per subplot
    if plot_num < 4:
        axes[plot_num].plot(x, y, label=f"h={h:.2f}")

# labeling each subplot graph
for k in range(4):
    axes[k].set_xlabel('x')
    axes[k].set_ylabel('y')
    axes[k].set_title(f"Euler's Method Runs {k*5 + 1} to {min((k+1)*5, len(h_vals))}")
    axes[k].legend()

    
plt.suptitle("Plot 1: euler Method on each h")
plt.tight_layout()
plt.show()


plt.figure(figsize=(8, 5))
plt.plot(h_vals, avg_err, 'bo-')
plt.xlabel('Step size h')
plt.ylabel('avg abs err')
plt.title('Plot 2: Avf err vs h')
plt.show()



plt.figure(figsize=(8, 5))
plt.plot(h_vals, local_err_15, 'ro-')
plt.xlabel('Step size h')
plt.ylabel('Local error at x = 15')
plt.title('Plot 3: Local Error at x = 15 vs h')
plt.show()
```
