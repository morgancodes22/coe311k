Using Ulers method to solve dy/dx =-2y*cos(x), y(0) = 1

```python
import numpy as np
import matplotlib.pyplot as plt


# Define parameters
h_vals = np.arange(0.01, 1.00, 0.05) # step size
#print(len(h_vals)) #debug
x_start, x_end = 0, 10*np.pi

# Function representing dy/dx = -2y
def f(x, y):
    return -2 * y * np.cos(x)
def exact(x,y):
    return 2*x+3*y

def Euler( y_not, step, start, stop):
    
    #setup
    num_steps = int((stop - start) / step)
    x = np.linspace(start, stop, num_steps + 1)
    y = np.zeros(num_steps + 1)
    
    #intial condition
    x[start] = start
    y[start] = y_not
    
    
    #go through each step
    for i in range(num_steps):
        x[i+ 1] = x[i] + step
        y[i + 1] = y[i] + h * f(x[i], y[i])
    return x,y


#subplots to keep organized
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

# Implement Euler's method for each h
for j in range(len(h_vals)): 
    h = h_vals[j]
    x,y = Euler(1 ,h ,0,10*np.pi)
    if j == 0:
        first_x = x.copy()
        first_y = y.copy()
        
    #compare current run to first run - need to match up the x values to be able to compare
    first_y_interp = np.interp(x, first_x, first_y) 
    
    globalError = np.abs(y - first_y_interp)
    
    
    #choose what subplot it goes on
    plot_num = j // 5 #5 lines per subplot
    axes[plot_num].plot(x, y, label=f"h={h:.2f}")
    
#labeling each subplot gragh
for k in range(4):
    axes[k].set_xlabel('x')
    axes[k].set_ylabel('y')
    axes[k].set_title(f"Euler's Method Runs {k*5 + 1} to {min((k+1)*5, len(h_vals))}")
    axes[k].legend()
    axes[k].grid(True)
    

plt.show()
```
