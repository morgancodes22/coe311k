# Plot monte carlo pi

```python
import numpy as np
import matplotlib.pyplot as plt 
import random
import math
```

### function

```python
def estPi(numPoints):
    
    count = 0 #counts what lands in the circle
    for i in range(0,numPoints):
        
        #get a random point
        x = random.uniform(-1,1)
        y = random.uniform(-1,1)
        
        #is it in the circle?
        if(x**2 + y**2 <1):
            count= count+1
    
    return 4.0*count/numPoints
```

### main

```python
x=np.arange(10,10000,20) #geneorate values

#run each value through a funcrion
pix = np.empty(len(x)) 
for i in range(0,len(x)):
    pix[i] = estPi(x[i])
#print(pix) #debug


error = np.round(abs(pix-np.pi), 15)
#print(error) #debug
```
