```python
import numpy as np
import matplotlib.pyplot as plt 
import random
import math 
```

make a function for the taylor series that allows changes to the number of terms

```python
def taylor(numTerms, x):
    runningSum=0
    for n in range(0,numTerms):
        num= x**(2*n + 1)
        den= math.factorial(2*n +1)
        runningSum = runningSum + (-1)**n * (num/den)
    return runningSum
#test = taylor(3, 1)
#print(test)
```

testing the series we just made for 0-10 with a .1 step size at dirrerent number of terms

```python
x = np.arange(0,10,.1) #x values being tested

oneTerm = taylor(1,x)
twoTerm = taylor(2, x)
threeTerm = taylor(3, x)
fourTerm = taylor(4, x)
#print(threeTerm[10]) # at x = 1 - tested this previouly on calc, know it works!
```

looking at error as more terms are added, compairing to the exact value of sin that is known!

```python
oneTermError = np.abs(np.sin(x) - oneTerm)
twoTermError = np.abs(np.sin(x) - twoTerm)
threeTermError = np.abs(np.sin(x) - threeTerm)
fourTermError = np.abs(np.sin(x) - fourTerm)
```

Plotting resluts!

```python
plt.title("truncation error")
plt.plot(x, oneTermError, 'r', label='oneTerm error')
plt.plot(x, twoTermError, 'g', label='twoTerm error')
plt.plot(x, threeTermError, 'b', label='threeTerm error')
plt.plot(x, fourTermError, 'y', label='fourTerm error')
plt.plot(x, np.sin(x), 'b--', label='sin(x) ')
plt.legend()
plt.show()
```
